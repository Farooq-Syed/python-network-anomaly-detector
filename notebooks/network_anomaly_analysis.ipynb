# Network Anomaly Analysis

This notebook gives a small walkthrough of the sample labeled dataset used in the repository. It can be extended later for feature exploration on UNSW-NB15 or another public dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/labeled_research_sample.csv')
df.head()
```

```python
df['label'].value_counts()
```

```python
features = ['bytes_sent', 'bytes_received', 'packets', 'duration_seconds', 'failed_logins', 'unique_ports']
df.groupby('label')[features].mean().T
```

```python
colors = df['label'].map({'benign': '#1d3557', 'attack': '#c1121f'})
plt.figure(figsize=(8, 5))
plt.scatter(df['bytes_sent'], df['packets'], c=colors)
plt.xlabel('bytes_sent')
plt.ylabel('packets')
plt.title('Bytes Sent vs Packets')
plt.show()
```
